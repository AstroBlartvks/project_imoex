```python
import requests
import json
import time
```

```python
get_url = lambda fr, to: f"https://iss.moex.com/iss/history/engines/stock/markets/index/boards/SNDX/securities/IMOEX.json?from={fr}&till={to}"
```

```python
def parse_data(fr: str, to: str):
    """fr, to - даты формата YYYY-mm-dd"""
    url: str = get_url(fr, to)

    json_data = requests.get(url)
    data = json.loads(json_data.text)

    return data['history']['data'][1:], data['history']['data'][-1][2]
```

```python
all_data = []

def start_parse_data(fr, to):
    global all_data

    try:
        while fr < to:
            data, fr = parse_data(fr, to)
            all_data += data
    except Exception as exp:
        time.sleep(3)
        print(str(exp), fr, to)
```

```python
start_parse_data("2010-00-00", "2026-04-04")
```

```python
json_data_all = json.dumps( all_data,
                            indent=4,
                            ensure_ascii=False,)
```
