## Подготовка данных для нейронок

```python
train_mask = df['TRADEDATE'] < '2020-01-01'
val_mask = (df['TRADEDATE'] >= '2020-01-01') & (df['TRADEDATE'] < '2022-01-01')
test_mask = df['TRADEDATE'] >= '2022-01-01'

print(f"Train: {train_mask.sum()} дней (2010–2019)")
print(f"Val:   {val_mask.sum()} дней (2020–2021)")
print(f"Test:  {test_mask.sum()} дней (2022–2026)")
```

```python
# Реализованная волатильность
# returns: df['LogReturn'] * 100

# Ground truth variance как в статье GINN:
# σ²_t = (r_t − μ̂_t)²
# где μ̂_t — скользящее среднее за 90 дней (AR-оценка)

rolling_mean = returns.rolling(window=90).mean()
raw_var = (returns - rolling_mean) ** 2

ground_truth_var = raw_var.rolling(window=5).mean()

valid_idx = ground_truth_var.dropna().index
print(f"Валидных точек после rolling: {len(valid_idx)}")
```

```python
# Формируем окна для нейронок
WINDOW = 90 

gt_var = ground_truth_var.dropna().values  # ground truth variance
gt_var = np.log1p(gt_var) # Волатильность имеет тяжёлый правый хвост (бывают редкие огромные значения). Log сжимает диапазон
trade_dates = df.loc[ground_truth_var.dropna().index, 'TRADEDATE'].values

X_windows = []
y_targets = []
window_dates = []

for i in range(WINDOW, len(gt_var)):
    X_windows.append(gt_var[i - WINDOW : i])  # 90 дней σ²
    y_targets.append(gt_var[i])                # σ² на следующий день
    window_dates.append(trade_dates[i])

X_windows = np.array(X_windows)  # shape: (N, 90)
y_targets = np.array(y_targets)  # shape: (N,)
window_dates = np.array(window_dates)

print(f"X shape: {X_windows.shape}")
print(f"y shape: {y_targets.shape}")
print(f"Период: {window_dates[0]} - {window_dates[-1]}")
```

```python
window_dates_dt = window_dates.astype('datetime64')

train_idx = window_dates_dt < np.datetime64('2020-01-01')
val_idx = (window_dates_dt >= np.datetime64('2020-01-01')) & (window_dates_dt < np.datetime64('2022-01-01'))
test_idx = window_dates_dt >= np.datetime64('2022-01-01')

X_train, y_train = X_windows[train_idx], y_targets[train_idx]
X_val, y_val = X_windows[val_idx], y_targets[val_idx]
X_test, y_test = X_windows[test_idx], y_targets[test_idx]
dates_test = window_dates[test_idx]  # оригинальные строки, если нужны

print(f"Train: {X_train.shape[0]} окон")
print(f"Val:   {X_val.shape[0]} окон")
print(f"Test:  {X_test.shape[0]} окон")
```

```python
# === Нормализация ===
train_mean = X_train.mean()
train_std = X_train.std()

X_train_norm = (X_train - train_mean) / train_std
X_val_norm = (X_val - train_mean) / train_std
X_test_norm = (X_test - train_mean) / train_std

y_train_norm = (y_train - train_mean) / train_std
y_val_norm = (y_val - train_mean) / train_std
y_test_norm = (y_test - train_mean) / train_std

print(f"Train mean: {train_mean:.6f}, std: {train_std:.6f}")
```

```python
class VolatilityDataset(torch.utils.data.Dataset):
    def __init__(self, X, y):
        # X shape: (N, 90) -> (N, 90, 1) для LSTM
        self.X = torch.FloatTensor(X).unsqueeze(-1)  # (N, 90, 1)
        self.y = torch.FloatTensor(y)                 # (N,)
    
    def __len__(self):
        return len(self.y)
    
    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]

train_dataset = VolatilityDataset(X_train_norm, y_train_norm)
val_dataset = VolatilityDataset(X_val_norm, y_val_norm)
test_dataset = VolatilityDataset(X_test_norm, y_test_norm)

BATCH_SIZE = 128

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False)

print(f"Train batches: {len(train_loader)}")
print(f"Val batches:   {len(val_loader)}")
print(f"Test batches:  {len(test_loader)}")

# Проверка одного батча
X_batch, y_batch = next(iter(train_loader))
print(f"\nBatch X shape: {X_batch.shape}")   # (64, 90, 1)
print(f"Batch y shape: {y_batch.shape}")     # (64,)
```

### Создание и обучение LSTM

```python
class VolatilityLSTM(nn.Module):
    """
    LSTM модель для предсказания волатильности
    Архитектура из статьи:
    - 3 LSTM слоя с шириной 256
    - Dropout слои между LSTM слоями
    - 2 линейных слоя (fully connected)
    - 1 BatchNorm слой
    - 1 ReLU активация
    """
    def __init__(self, input_size=1, hidden_size=256, num_layers=3, dropout=0.2):
        super(VolatilityLSTM, self).__init__()
        
        self.hidden_size = hidden_size
        self.num_layers = num_layers
        
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            dropout=dropout if num_layers > 1 else 0,
            batch_first=True
        )
        
        self.batch_norm = nn.BatchNorm1d(hidden_size)
        
        self.fc1 = nn.Linear(hidden_size, 128)
        self.fc2 = nn.Linear(128, 1)
        
        self.dropout = nn.Dropout(dropout)
        
        self.relu = nn.ReLU()
        
    def forward(self, x):
        # x shape: (batch_size, sequence_length, input_size)
        
        lstm_out, (hidden, cell) = self.lstm(x)
        # lstm_out: (batch_size, sequence_length, hidden_size)
        
        last_output = lstm_out[:, -1, :]  # (batch_size, hidden_size)
        normalized = self.batch_norm(last_output)
        
        fc1_out = self.relu(self.fc1(normalized))
        fc1_out = self.dropout(fc1_out)
        
        output = self.fc2(fc1_out)
        
        return output.squeeze(-1)
```

```python
LSTM_EPOCHS = 300
lstm_model = VolatilityLSTM(hidden_size=128, num_layers=3)
criterion = nn.MSELoss() #QLIKELoss(train_mean, train_std) LSTM Test: MSE=487.8656, MAE=2.2646, R²=0.3309
optimizer = optim.AdamW(lstm_model.parameters(), lr=1e-3)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', patience=20, factor=0.5)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
lstm_model.to(device)
```

```python
best_val_loss = float('inf')
best_model_state = None
patience_counter = 0
PATIENCE = 100 #Увеличил - стало лучше!

for epoch in range(LSTM_EPOCHS):
    lstm_model.train()
    train_losses = []
    
    for X_batch, y_batch in train_loader:
        optimizer.zero_grad()
        outputs = lstm_model(X_batch.to(device))
        loss = criterion(outputs, y_batch.to(device))
        loss.backward()
        torch.nn.utils.clip_grad_norm_(lstm_model.parameters(), max_norm=1.0)
        optimizer.step()
        
        train_losses.append(loss.item())
    
    avg_train_loss = np.mean(train_losses)
    
    lstm_model.eval()
    val_losses = []
    
    with torch.no_grad():
        for X_batch, y_batch in val_loader:
            outputs = lstm_model(X_batch.to(device))
            loss = criterion(outputs, y_batch.to(device))
            val_losses.append(loss.item())
    
    avg_val_loss = np.mean(val_losses)
    scheduler.step(avg_val_loss)

    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        best_model_state = lstm_model.state_dict().copy()
        patience_counter = 0
    else:
        patience_counter += 1
    
    if patience_counter >= PATIENCE:
        print(f"Ранняя остановка на эпохе {epoch+1}")
        break

    if (epoch + 1) % 25 == 0:
        current_lr = optimizer.param_groups[0]['lr']
        print(f"Измененный LR: {current_lr:.6f}")
    
    if (epoch + 1) % 10 == 0 or epoch == 0:
        print(f"Эпоха {epoch+1}/{LSTM_EPOCHS} - Train Loss: {avg_train_loss:.6f}, Val Loss: {avg_val_loss:.6f}")


lstm_model.load_state_dict(best_model_state)
```

```python
lstm_model.eval()
all_preds = []
all_targets = []

with torch.no_grad():
    for X_batch, y_batch in test_loader:
        preds = lstm_model(X_batch.to(device)).cpu().numpy()
        all_preds.extend(preds)
        all_targets.extend(y_batch.numpy())

all_preds = np.array(all_preds)
all_targets = np.array(all_targets)

# Денормализация
preds_denorm = np.expm1(all_preds * train_std + train_mean)
targets_denorm = np.expm1(all_targets * train_std + train_mean)

mse = mean_squared_error(targets_denorm, preds_denorm)
mae = mean_absolute_error(targets_denorm, preds_denorm)
r2 = r2_score(targets_denorm, preds_denorm)

print(f"LSTM Test: MSE={mse:.4f}, MAE={mae:.4f}, R²={r2:.4f}")
```

## модель Transformer

```python
class VolatilityTransformer(nn.Module):
    def __init__(self, input_size=1, d_model=64, nhead=4, num_layers=2, dropout=0.2, seq_len=90):
        super().__init__()
        
        self.embedding = nn.Linear(input_size, d_model)
        self.pos_encoding = nn.Parameter(torch.randn(1, seq_len, d_model) * 0.01)
        
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=d_model * 4,
            dropout=dropout,
            batch_first=True
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
        
        self.head = nn.Sequential(
            nn.Linear(d_model, d_model),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(d_model, 1)
        )
    
    def forward(self, x):
        # x: (batch, 90, 1)
        x = self.embedding(x) + self.pos_encoding  # (batch, 90, d_model)
        x = self.transformer(x)                      # (batch, 90, d_model)
        x = x[:, -1, :]                              # берём последний timestep
        return self.head(x).squeeze(-1)               # (batch,)
```

```python
TRANS_EPOCHS = 300
trans_model = VolatilityTransformer(d_model=64, num_layers=2)
trans_criterion = nn.MSELoss()
trans_optimizer = optim.AdamW(trans_model.parameters(), lr=1e-3)
trans_scheduler = optim.lr_scheduler.ReduceLROnPlateau(trans_optimizer, mode='min', patience=20, factor=0.5)

trans_model.to(device)

best_val_loss = float('inf')
best_model_state = None
patience_counter = 0
PATIENCE = 100

for epoch in range(TRANS_EPOCHS):
    trans_model.train()
    train_losses = []
    
    for X_batch, y_batch in train_loader:
        trans_optimizer.zero_grad()
        outputs = trans_model(X_batch.to(device))
        loss = trans_criterion(outputs, y_batch.to(device))
        loss.backward()
        torch.nn.utils.clip_grad_norm_(trans_model.parameters(), max_norm=1.0)
        trans_optimizer.step()
        train_losses.append(loss.item())
    
    avg_train_loss = np.mean(train_losses)
    
    trans_model.eval()
    val_losses = []
    with torch.no_grad():
        for X_batch, y_batch in val_loader:
            preds = trans_model(X_batch.to(device))
            loss = trans_criterion(preds, y_batch.to(device))
            val_losses.append(loss.item())
    
    avg_val_loss = np.mean(val_losses)
    trans_scheduler.step(avg_val_loss)
    
    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        best_model_state = {k: v.clone() for k, v in trans_model.state_dict().items()}
        patience_counter = 0
    else:
        patience_counter += 1
    
    if (epoch + 1) % 10 == 0:
        print(f"Эпоха {epoch+1}/{TRANS_EPOCHS} - Train: {avg_train_loss:.6f}, Val: {avg_val_loss:.6f}")
    
    if patience_counter >= PATIENCE:
        print(f"Ранняя остановка на эпохе {epoch+1}")
        break

trans_model.load_state_dict(best_model_state)
```

```python
trans_model.eval()
trans_preds = []

with torch.no_grad():
    for X_batch, y_batch in test_loader:
        preds = trans_model(X_batch.to(device)).cpu().numpy()
        trans_preds.extend(preds)

trans_preds = np.array(trans_preds)
trans_preds_denorm = np.expm1(trans_preds * train_std + train_mean)

trans_mse = mean_squared_error(targets_denorm, trans_preds_denorm)
trans_mae = mean_absolute_error(targets_denorm, trans_preds_denorm)
trans_r2 = r2_score(targets_denorm, trans_preds_denorm)

print(f"{'Модель':<15} {'MSE':>10} {'MAE':>10} {'R²':>10}")
print(f"{'GARCH':<15} {garch_mse:>10.4f} {garch_mae:>10.4f} {garch_r2:>10.4f}")
print(f"{'LSTM':<15} {mse:>10.4f} {mae:>10.4f} {r2:>10.4f}")
print(f"{'Transformer':<15} {trans_mse:>10.4f} {trans_mae:>10.4f} {trans_r2:>10.4f}")
```

## GINN (GARCH-Informed Neural Network)

Гибрид из статьи Xu et al. (2024, Carnegie Mellon).
Та же LSTM архитектура, но loss = λ·MSE(real, pred) + (1-λ)·MSE(garch, pred).

```python
# GARCH rolling forecast для ВСЕХ окон (train + val + test)
# Это нужно чтобы GINN мог учиться у GARCH

print("Считаю GARCH предсказания для всех окон ")

garch_all_preds = []

for i in range(len(X_windows)):
    # Позиция в returns: WINDOW (90) + offset от rolling + i
    pos = ground_truth_var.dropna().index[WINDOW + i]
    train_data = returns.iloc[pos - 90 : pos]
    
    try:
        m = arch_model(train_data, vol='Garch', p=1, q=1, dist='Normal')
        r = m.fit(disp='off')
        f = r.forecast(horizon=1)
        garch_all_preds.append(f.variance.values[-1, 0])
    except:
        garch_all_preds.append(np.nan)
    
    if (i + 1) % 500 == 0:
        print(f"  {i+1}/{len(X_windows)} готово")

garch_all_preds = np.array(garch_all_preds)

nan_mask = np.isnan(garch_all_preds)
if nan_mask.sum() > 0:
    garch_all_preds[nan_mask] = np.nanmean(garch_all_preds)
    print(f"  Заполнено {nan_mask.sum()} NaN")

print(f"Готово: {len(garch_all_preds)} GARCH предсказаний")
```

```python
# Нормализуем GARCH своими параметрами
# garch_log = np.log1p(garch_all_preds)
# garch_mean = garch_log.mean()
# garch_std = garch_log.std()

# garch_all_norm = (garch_log - garch_mean) / garch_std

garch_log = np.log1p(garch_all_preds)
garch_all_norm = (garch_log - train_mean) / train_std   # ← те же train_mean/std, что и для y!

# А для y используем train_mean/train_std как раньше
y_train_norm = (y_train - train_mean) / train_std

# Split по тем же индексам
garch_train_norm = garch_all_norm[train_idx]
garch_val_norm = garch_all_norm[val_idx]
garch_test_norm = garch_all_norm[test_idx]

print(f"GARCH norm — train: {len(garch_train_norm)}, val: {len(garch_val_norm)}, test: {len(garch_test_norm)}")
```

```python
# Dataset для GINN: X, y_real, y_garch

class GINNDataset(torch.utils.data.Dataset):
    def __init__(self, X, y_real, y_garch):
        self.X = torch.FloatTensor(X).unsqueeze(-1)  # (N, 90, 1)
        self.y_real = torch.FloatTensor(y_real)
        self.y_garch = torch.FloatTensor(y_garch)
    
    def __len__(self):
        return len(self.y_real)
    
    def __getitem__(self, idx):
        return self.X[idx], self.y_real[idx], self.y_garch[idx]

ginn_train = GINNDataset(X_train_norm, y_train_norm, garch_train_norm)
ginn_val = GINNDataset(X_val_norm, y_val_norm, garch_val_norm)
ginn_test = GINNDataset(X_test_norm, y_test_norm, garch_test_norm)

ginn_train_loader = DataLoader(ginn_train, batch_size=512, shuffle=True)
ginn_val_loader = DataLoader(ginn_val, batch_size=512, shuffle=False)
ginn_test_loader = DataLoader(ginn_test, batch_size=512, shuffle=False)
```

```python
ginn_model = VolatilityLSTM(hidden_size=128, num_layers=3)
ginn_model.to(device)

ginn_optimizer = optim.AdamW(ginn_model.parameters(), lr=1e-3)
ginn_scheduler = optim.lr_scheduler.ReduceLROnPlateau(ginn_optimizer, mode='min', patience=20, factor=0.5)
mse_loss = nn.MSELoss()
```

```python
# === GINN с меньшей моделью ===
LAMBDA = 0.7

best_val_loss = float('inf')
best_model_state = None
patience_counter = 0
PATIENCE = 100  # больше терпения

VAL_LOSSES = []
TRN_LOSSES = []

for epoch in range(300):
    ginn_model.train()
    train_losses = []
    
    for X_batch, y_real_batch, y_garch_batch in ginn_train_loader:
        X_batch = X_batch.to(device)
        y_real_batch = y_real_batch.to(device)
        y_garch_batch = y_garch_batch.to(device)
        
        ginn_optimizer.zero_grad()
        preds = ginn_model(X_batch)
        
        loss = LAMBDA * mse_loss(preds, y_real_batch) + 2 * (1 - LAMBDA) * mse_loss(preds, y_garch_batch)
        
        loss.backward()
        torch.nn.utils.clip_grad_norm_(ginn_model.parameters(), max_norm=1.0)
        ginn_optimizer.step()
        train_losses.append(loss.item())
    
    avg_train = np.mean(train_losses)
    
    ginn_model.eval()
    val_losses = []
    with torch.no_grad():
        for X_batch, y_real_batch, y_garch_batch in ginn_val_loader:
            preds = ginn_model(X_batch.to(device))
            loss = LAMBDA * mse_loss(preds, y_real_batch.to(device)) + (1 - LAMBDA) * mse_loss(preds, y_garch_batch.to(device))
            val_losses.append(loss.item())
    
    avg_val = np.mean(val_losses)

    VAL_LOSSES.append(avg_val)
    TRN_LOSSES.append(avg_train)

    ginn_scheduler.step(avg_val)
    
    if avg_val < best_val_loss:
        best_val_loss = avg_val
        best_model_state = {k: v.clone() for k, v in ginn_model.state_dict().items()}
        patience_counter = 0
    else:
        patience_counter += 1
    
    if (epoch + 1) % 10 == 0:
        print(f"Эпоха {epoch+1}/500 - Train: {avg_train:.6f}, Val: {avg_val:.6f}")
    
    if patience_counter >= PATIENCE:
        print(f"Ранняя остановка на эпохе {epoch+1}")
        break

ginn_model.load_state_dict(best_model_state)
print("GINN обучен!")
```

```python
class FeatureGatingGINN(nn.Module):
    def __init__(self, input_size=1, hidden_size=128, num_layers=3, dropout=0.2):
        super().__init__()
        
        # Основной LSTM для данных
        self.data_lstm = nn.LSTM(input_size, hidden_size, num_layers, 
                                 batch_first=True, dropout=dropout)
        self.data_bn = nn.BatchNorm1d(hidden_size)
        
        # Проекция GARCH сигнала
        self.garch_proj = nn.Sequential(
            nn.Linear(1, hidden_size // 2),
            nn.ReLU(),
            nn.Linear(hidden_size // 2, hidden_size)
        )
        
        # Сеть для предсказания лямбды
        self.lambda_lstm = nn.LSTM(input_size, 32, 2, batch_first=True, bidirectional=True)
        self.lambda_bn = nn.BatchNorm1d(64)
        self.lambda_head = nn.Sequential(
            nn.Linear(64, 16),
            nn.ReLU(),
            nn.Linear(16, 1),
            nn.Sigmoid()
        )
        
        # Выходной слой
        self.output_head = nn.Sequential(
            nn.Linear(hidden_size, hidden_size // 2),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_size // 2, 1)
        )
        
    def forward(self, x, garch_signal):
        # 1. Обрабатываем исторические данные
        out, _ = self.data_lstm(x)
        data_feat = self.data_bn(out[:, -1, :])
        
        # 2. Проецируем GARCH сигнал
        garch_feat = self.garch_proj(garch_signal)
        
        # 3. Вычисляем лямбду
        l_out, _ = self.lambda_lstm(x)
        l_feat = self.lambda_bn(l_out[:, -1, :])
        lambda_val = self.lambda_head(l_feat)
        
        # 4. Комбинируем и предсказываем
        combined = data_feat + lambda_val * garch_feat
        output = self.output_head(combined).squeeze(-1)
        
        return output, lambda_val.squeeze(-1)
```

```python
class GatingDataset(Dataset):
    def __init__(self, X, y_real, y_garch):
        self.X = torch.FloatTensor(X).unsqueeze(-1)
        self.y_real = torch.FloatTensor(y_real)
        self.y_garch = torch.FloatTensor(y_garch)
        
    def __len__(self): return len(self.y_real)
    
    def __getitem__(self, idx):
        return self.X[idx], self.y_real[idx], self.y_garch[idx]
```

```python
model = FeatureGatingGINN(hidden_size=128).to(device)
```

```python
def train_gating_model(model, train_loader, val_loader, epochs=300, patience=50):
    # ВСЕ параметры в одной группе (проще и надёжнее)
    optimizer = optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, patience=20, factor=0.5)
    mse = nn.MSELoss()
    
    best_val = float('inf')
    best_state = None
    patience_cnt = 0
    history = {'loss': [], 'val_loss': [], 'lambda': [], 'val_lambda': []}
    
    for epoch in range(epochs):
        # --- Train ---
        model.train()
        tr_loss, tr_lam = [], []
        for X, y_r, y_g in train_loader:
            X, y_r, y_g = X.to(device), y_r.to(device), y_g.to(device)
            
            optimizer.zero_grad()
            pred, lam = model(X, y_g.unsqueeze(-1))
            
            # Основной лосс + небольшая регуляризация на лямбду
            reg = 0.001 * ((lam - 0.5)**2).mean()
            loss = mse(pred, y_r) + reg
            
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            
            tr_loss.append(loss.item())
            tr_lam.append(lam.mean().item())
        
        # --- Val ---
        model.eval()
        val_loss, val_lam = [], []
        with torch.no_grad():
            for X, y_r, y_g in val_loader:
                X, y_r, y_g = X.to(device), y_r.to(device), y_g.to(device)
                pred, lam = model(X, y_g.unsqueeze(-1))
                val_loss.append(mse(pred, y_r).item())
                val_lam.append(lam.mean().item())
        
        # Метрики эпохи
        avg_tr_loss = np.mean(tr_loss)
        avg_val_loss = np.mean(val_loss)
        avg_tr_lam = np.mean(tr_lam)
        avg_val_lam = np.mean(val_lam)
        
        history['loss'].append(avg_tr_loss)
        history['val_loss'].append(avg_val_loss)
        history['lambda'].append(avg_tr_lam)
        history['val_lambda'].append(avg_val_lam)
        
        scheduler.step(avg_val_loss)
        
        if avg_val_loss < best_val:
            best_val = avg_val_loss
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            patience_cnt = 0
        else:
            patience_cnt += 1
            
        if (epoch+1) % 20 == 0:
            print(f"Ep {epoch+1:3d} | TrLoss={avg_tr_loss:.4f} | ValLoss={avg_val_loss:.4f} | λ_tr={avg_tr_lam:.3f} | λ_val={avg_val_lam:.3f}")
            
        if patience_cnt >= patience:
            print(f"Early stop at epoch {epoch+1}")
            break
    
    model.load_state_dict(best_state)
    return model, history
```

```python
train_ds = GatingDataset(X_train_norm, y_train_norm, garch_train_norm)
val_ds   = GatingDataset(X_val_norm,   y_val_norm,   garch_val_norm)
test_ds  = GatingDataset(X_test_norm,  y_test_norm,  garch_test_norm)

train_loader = DataLoader(train_ds, batch_size=256, shuffle=True)
val_loader   = DataLoader(val_ds,   batch_size=128)
test_loader  = DataLoader(test_ds,  batch_size=128)

# Создаём и обучаем модель
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = FeatureGatingGINN().to(device)
print(f"Model params: {sum(p.numel() for p in model.parameters()):,}")

model, history = train_gating_model(model, train_loader, val_loader, epochs=300)
```

```python

# 1. Тестируем Feature Gating GINN
@torch.no_grad()
def test_gating_model(model, loader):
    """Тестирование модели с гейтингом"""
    model.eval()
    preds_list = []
    lambs_list = []
    with torch.no_grad():
        for X_batch, y_real, y_garch in loader:
            X_batch = X_batch.to(device)
            y_garch_batch = y_garch.to(device)
            preds, lambs = model(X_batch, y_garch_batch.unsqueeze(-1))
            preds_list.extend(preds.cpu().numpy())
            lambs_list.extend(lambs.cpu().numpy())
    
    preds_arr = np.array(preds_list)
    lambs_arr = np.array(lambs_list)
    preds_denormed = np.expm1(preds_arr * train_std + train_mean)
    
    return preds_denormed, lambs_arr


# Тестируем
gating_preds_denorm, gating_lambdas = test_gating_model(model, test_loader)

# Метрики
gating_mse = mean_squared_error(targets_denorm, gating_preds_denorm)
gating_mae = mean_absolute_error(targets_denorm, gating_preds_denorm)
gating_r2 = r2_score(targets_denorm, gating_preds_denorm)


# 2. Сводная таблица результатов
print(f"\n{'='*70}")
print(f"СВОДНАЯ ТАБЛИЦА РЕЗУЛЬТАТОВ")
print(f"{'='*70}")
print(f"{'Модель':<25} {'MSE':>12} {'MAE':>10} {'R²':>10}")
print(f"{'-'*57}")
print(f"{'GARCH(1,1)':<25} {garch_mse:>12.2f} {garch_mae:>10.2f} {garch_r2:>10.4f}")
print(f"{'LSTM (чистый)':<25} {mse:>12.2f} {mae:>10.2f} {r2:>10.4f}")
print(f"{'Transformer':<25} {trans_mse:>12.2f} {trans_mae:>10.2f} {trans_r2:>10.4f}")
print(f"{'GINN Adaptive Gate':<25} {gating_mse:>12.2f} {gating_mae:>10.2f} {gating_r2:>10.4f}")
print(f"{'='*70}")

# Сравнение с лучшими
improvement_vs_garch = (1 - gating_mse/garch_mse) * 100
improvement_vs_lstm = (1 - gating_mse/mse) * 100

print(f"\nУлучшение относительно GARCH: {improvement_vs_garch:+.1f}%")
print(f"Улучшение относительно LSTM:  {improvement_vs_lstm:+.1f}%")
print(f"Среднее λ на тесте: {np.mean(gating_lambdas):.4f} ± {np.std(gating_lambdas):.4f}")


# 3. Детальная визуализация
fig = make_subplots(
    rows=4, cols=1,
    subplot_titles=(
        'Все модели: предсказанная волатильность',
        'GINN Adaptive Gate: детальный график',
        'Адаптивный λ(x) по времени',
        'Распределение λ на тесте'
    ),
    vertical_spacing=0.08,
    row_heights=[0.35, 0.25, 0.20, 0.20]
)

# === Row 1: Все модели ===
fig.add_trace(
    go.Scatter(x=dates_test, y=targets_denorm, name='Realized σ²',
               line=dict(color='black', width=1.5), opacity=0.5),
    row=1, col=1
)
fig.add_trace(
    go.Scatter(x=dates_test, y=garch_preds_raw, name='GARCH(1,1)',
               line=dict(color='red', width=1.2), opacity=0.6),
    row=1, col=1
)
fig.add_trace(
    go.Scatter(x=dates_test, y=preds_denorm, name='LSTM',
               line=dict(color='orange', width=1.2), opacity=0.6),
    row=1, col=1
)
fig.add_trace(
    go.Scatter(x=dates_test, y=gating_preds_denorm, name='GINN Gate',
               line=dict(color='blue', width=2), opacity=0.9),
    row=1, col=1
)

# === Row 2: GINN детально ===
fig.add_trace(
    go.Scatter(x=dates_test, y=targets_denorm, name='Реальность',
               line=dict(color='black', width=1), opacity=0.4),
    row=2, col=1
)
fig.add_trace(
    go.Scatter(x=dates_test, y=gating_preds_denorm, name='GINN Gate',
               line=dict(color='blue', width=2), 
               fill='tonexty', fillcolor='rgba(0,0,255,0.1)'),
    row=2, col=1
)

# === Row 3: λ по времени ===
fig.add_trace(
    go.Scatter(x=dates_test, y=gating_lambdas, name='λ(x)',
               line=dict(color='purple', width=1.5),
               fill='tozeroy', fillcolor='rgba(128,0,128,0.15)'),
    row=3, col=1
)
fig.add_hline(y=0.5, line=dict(dash='dash', color='green', width=1.5), 
              annotation_text="Баланс λ=0.5", row=3, col=1)

# Добавляем зоны интерпретации
fig.add_hrect(y0=0, y1=0.3, fillcolor="red", opacity=0.08,
              layer="below", line_width=0, row=3, col=1)
fig.add_hrect(y0=0.7, y1=1.0, fillcolor="blue", opacity=0.08, 
              layer="below", line_width=0, row=3, col=1)

fig.add_annotation(xref="paper", yref="y3", x=0.01, y=0.15, 
                   text="⇒ GARCH", showarrow=False, font=dict(color="red"), row=3, col=1)
fig.add_annotation(xref="paper", yref="y3", x=0.01, y=0.85, 
                   text="⇒ Data", showarrow=False, font=dict(color="blue"), row=3, col=1)

# === Row 4: Гистограмма λ ===
fig.add_trace(
    go.Histogram(x=gating_lambdas, nbinsx=40, name='Распределение λ',
                 marker_color='purple', opacity=0.7,
                 hovertemplate='λ=%{x:.3f}<br>Count=%{y}<extra></extra>'),
    row=4, col=1
)
fig.add_vline(x=np.mean(gating_lambdas), line=dict(color='purple', width=2, dash='dash'),
              annotation_text=f"μ={np.mean(gating_lambdas):.3f}", row=4, col=1)
fig.add_vline(x=0.5, line=dict(color='green', width=1.5, dash='dot'), row=4, col=1)

# Настройка осей
fig.update_yaxes(title_text="σ² variance", row=1, col=1)
fig.update_yaxes(title_text="σ² variance", row=2, col=1)
fig.update_yaxes(title_text="λ", range=[0, 1], row=3, col=1)
fig.update_yaxes(title_text="Частота", row=4, col=1)
fig.update_xaxes(title_text="Дата", row=3, col=1)
fig.update_xaxes(title_text="λ", row=4, col=1)

fig.update_layout(
    title=f'GINN Feature Gating — Adaptive λ (MSE={gating_mse:.1f}, R²={gating_r2:.3f}, λ̄={np.mean(gating_lambdas):.3f})',
    height=1200,
    hovermode='x unified',
    showlegend=True,
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="right",
        x=1
    )
)

fig.write_html("./plots/gin_adaptive_gate_final.html")
fig.show()


# 4. Анализ режимов работы λ
print(f"\n{'='*70}")
print(f"АНАЛИЗ РЕЖИМОВ РАБОТЫ АДАПТИВНОГО λ")
print(f"{'='*70}")

# Разбиваем на три режима
low_mask = gating_lambdas < 0.3      # доверяем GARCH
mid_mask = (gating_lambdas >= 0.3) & (gating_lambdas <= 0.7)  # баланс
high_mask = gating_lambdas > 0.7     # доверяем данным

for regime_name, mask in [("GARCH-heavy (λ<0.3)", low_mask), 
                           ("Balanced (0.3≤λ≤0.7)", mid_mask),
                           ("Data-driven (λ>0.7)", high_mask)]:
    if mask.sum() > 0:
        print(f"\n{regime_name}:")
        print(f"  Дней: {mask.sum()} ({100*mask.mean():.1f}%)")
        print(f"  Средняя σ²: {np.mean(targets_denorm[mask]):.2f}")
        print(f"  Медианная σ²: {np.median(targets_denorm[mask]):.2f}")
        print(f"  Средняя ошибка GINN: {np.mean(np.abs(targets_denorm[mask] - gating_preds_denorm[mask])):.2f}")
        print(f"  Средняя ошибка GARCH: {np.mean(np.abs(targets_denorm[mask] - garch_preds_raw[mask])):.2f}")

# Сравнение ошибок по режимам
print(f"\nПревосходство GINN над GARCH по режимам:")
for regime_name, mask in [("GARCH-heavy", low_mask), 
                           ("Balanced", mid_mask),
                           ("Data-driven", high_mask)]:
    if mask.sum() > 0:
        ginn_err = np.abs(targets_denorm[mask] - gating_preds_denorm[mask]).mean()
        garch_err = np.abs(targets_denorm[mask] - garch_preds_raw[mask]).mean()
        improvement = (1 - ginn_err/garch_err) * 100
        print(f"  {regime_name}: GINN ошибка={ginn_err:.2f} vs GARCH={garch_err:.2f} ({improvement:+.1f}%)")
```
